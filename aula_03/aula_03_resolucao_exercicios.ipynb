### Exercício 03 · Qual será a saída?

Leia o código linha a linha e escreva o que será mostrado. Explique por que o valor final de `horas` não é 4. Só depois execute a célula.

````python
horas = 4
horas = horas + 2
horas = horas * 3
print(f"Uso diário: {horas} h")
````

<details><summary>Conferir a resposta</summary>

A saída é `Uso diário: 18 h`.
</details>

<details><summary>Por que esta é a resposta?</summary>

| Linha | O que Python calcula | `horas` passa a valer |
| --- | --- | --- |
| `horas = 4` | Nada: só guarda o valor | 4 |
| `horas = horas + 2` | 4 + 2 | 6 |
| `horas = horas * 3` | 6 × 3 | 18 |

O sinal `=` não indica igualdade. Python calcula primeiro o lado direito, com o valor atual, e depois **substitui** o que estava guardado.

O valor final não é 4 porque o 4 só existiu até a segunda linha. O `print()` mostra o valor de `horas` no momento em que é executado: 18.
</details>

````python
# Escreva sua solução aqui.
````

### Exercício 08 · Quantas voltas o laço dá?

Sem executar, escreva a saída de cada trecho. Depois, confira na célula abaixo.

````python
for i in range(1, 6):
    print(i * i)

for n in range(3, 8):
    if n == 6:
        break
    print(n)
````

<details><summary>Conferir a resposta</summary>

O primeiro trecho mostra `1`, `4`, `9`, `16` e `25`. O segundo mostra `3`, `4` e `5`.
</details>

<details><summary>Por que esta é a resposta?</summary>

- **Primeiro trecho:** o fim do `range` não entra, então `i` vai de 1 a 5, e não até 6. São cinco voltas, e cada uma mostra o quadrado de `i`.
- **Segundo trecho:** com `n` valendo 3, 4 e 5, a condição é falsa e o `print` é executado.
- **Por que o 6 não aparece:** o teste vem antes do `print`. Se o `print` viesse antes do `if`, o 6 apareceria.
- **Por que o 7 não aparece:** `break` encerra todo o laço, e não apenas aquela volta.
</details>

### Exercício 12 · Quantos seis saíram?

Lance um dado de seis faces **20 vezes** com `random.randint(1, 6)`. Mostre cada resultado e, no final, quantas vezes saiu o número 6.

**Pense antes de programar:** responda em português, sem escrever código.

1. O que acontece uma vez só? E o que se repete?
2. Que valor o contador deve ter antes do laço? Quando ele aumenta?

<details><summary>Conferir as respostas</summary>

1. Uma vez: criar o contador, antes, e mostrar o total, depois. Repete: sortear, mostrar e conferir se saiu 6.
2. Começa em zero e aumenta um ponto somente quando o dado dá 6.
</details>

<details><summary>Ver os passos para programar</summary>

1. Importe `random` e crie o contador com zero, antes do `for`.
2. Repita 20 vezes: sorteie o dado e mostre o valor.
3. Ainda dentro do `for`, some um ao contador quando o dado for 6.
4. Depois do `for`, mostre o total.
</details>

````python
# Escreva sua solução aqui.
````

<details><summary>Ver uma solução</summary>

```python
import random

seis = 0
for lancamento in range(1, 21):
    dado = random.randint(1, 6)
    print(f"Lançamento {lancamento}: {dado}")
    if dado == 6:
        seis = seis + 1

print(f"Total de seis: {seis}")
```

Os valores sorteados podem variar a cada execução. Conte os seis na saída e confira se o total bate.
</details>

<details><summary>Por que esta é a resposta?</summary>

- **`seis = 0` antes do laço:** dentro dele, o contador voltaria a zero a cada volta, e o total final só poderia ser 0 ou 1.
- **`randint` dentro do laço:** cada lançamento precisa de um novo sorteio. Fora dele, o mesmo valor apareceria 20 vezes.
- **`range(1, 21)`:** produz de 1 a 20, ou seja, 20 lançamentos.
- **`if dado == 6`:** o contador só aumenta quando sai 6, e não a cada volta.
- **Último `print()` sem indentação:** ele fica fora do `for`, por isso o total aparece uma vez, no final.
</details>
