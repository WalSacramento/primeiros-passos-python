## 2. Variáveis e tipos de dados

Uma variável guarda um valor que pode ser usado em outras células. O nome descreve o que o valor representa. Em Python, o tipo pertence ao **valor** e não precisa ser declarado antes.

- `int`: número inteiro, como `220`
- `float`: número decimal, como `0.75` (**use ponto**, não vírgula)
- `str`: texto, como `"lâmpada"`
- `bool`: verdadeiro (`True`) ou falso (`False`)

````python
tensao = 220
corrente = 0.75
aparelho = "lâmpada"
ligado = True

print(aparelho, tensao, corrente, ligado)
print(type(tensao), type(corrente), type(aparelho), type(ligado))
````

## 3. Operações e saída formatada

Python usa `+`, `-`, `*` e `/` para as quatro operações. Aqui, a potência elétrica é `P = V × I`. Uma *f-string* coloca os valores dentro da mensagem; `:.2f` mostra duas casas decimais. Python apresenta esses decimais com ponto.

````python
potencia = tensao * corrente
print(f"Potência estimada: {potencia:.2f} W")
print(f"Metade da potência: {potencia / 2:.2f} W")
````

## 4. Entrada de dados

`input()` lê o que foi digitado e devolve **texto**, mesmo quando parece um número. Use `float()` para converter um decimal ou `int()` para converter um inteiro. Ao executar a próxima célula, digite um número usando ponto decimal, por exemplo `100.5`.

````python
texto_potencia = input("Potência do aparelho em W: ")
potencia_digitada = float(texto_potencia)
print(f"Você informou {potencia_digitada:.2f} W")
````

## 5. Decisões com `if` e `else`

Uma comparação produz `True` ou `False`. O bloco indentado abaixo de `if` executa quando a condição é verdadeira; `else` cobre o outro caso. `>` significa *maior que* e `<=` significa *menor ou igual*. A indentação faz parte da sintaxe.

````python
limite_w = 100
print(potencia_digitada > limite_w)

if potencia_digitada > limite_w:
    print("Potência acima de 100 W")
else:
    print("Potência de até 100 W")
````

## 6. Projeto guiado: estimador de consumo de energia

**Situação:** queremos estimar a energia consumida por um aparelho e seu custo no período. Informaremos potência em **W**, horas de uso **por dia**, número de **dias**, tarifa em **R$/kWh** e um orçamento em **R$**.

A energia é `potência (W) ÷ 1000 × horas por dia × dias`, em **kWh**. O custo é `energia (kWh) × tarifa (R$/kWh)`. Trata-se de uma estimativa: impostos, bandeiras e variação real de potência não entram na conta.

Para esta aula, informe potência e dias **maiores que zero**, horas por dia **entre 0 e 24**, e tarifa e orçamento **não negativos**. Se digitar algo fora dessas condições, corrija a entrada e execute a célula novamente.

### Etapa 1 — Dados e energia (7 min)

**Sua vez:** use `100 W`, `5 h/dia` e `30 dias`. Divida a potência por `1000` e calcule a energia. **Resultado esperado: 15 kWh.**

````python
potencia_w = 100
horas_dia = 5
dias = 30
# Calcule energia_kwh e mostre o resultado.
````

````python
# Solução da etapa 1
energia_kwh = potencia_w / 1000 * horas_dia * dias
print(f"Consumo: {energia_kwh:.2f} kWh")
````

### Etapa 2 — Custo (5 min)

**Sua vez:** com tarifa de `R$ 0,80/kWh`, calcule o custo para os `15 kWh`. **Resultado esperado: R$ 12,00.**

````python
tarifa = 0.80
# Calcule custo e mostre o resultado.
````

````python
# Solução da etapa 2
custo = energia_kwh * tarifa
print(f"Custo estimado: R$ {custo:.2f}")
````

### Etapa 3 — Decisão (5 min)

**Sua vez:** compare o custo com um orçamento de `R$ 10,00`. Mostre uma mensagem quando o custo ultrapassar o orçamento e outra caso contrário. **Resultado esperado:** acima do orçamento.

````python
orcamento = 10.00
# Escreva aqui if e else para comparar custo com orcamento.
````

````python
# Solução da etapa 3
if custo > orcamento:
    print("Custo acima do orçamento")
else:
    print("Custo dentro do orçamento")
````

### Etapa 4 — Programa completo com entrada (8 min)

**Sua vez:** reúna entrada, conversão, cálculos, saída e decisão. Execute a solução abaixo primeiro com `100`, `5`, `30`, `0.80`, `10`; depois, com orçamento `15`. Digite números com ponto decimal.

````python
# Solução completa: execute e responda às perguntas.
potencia_w = float(input("Potência do aparelho (W): "))
horas_dia = float(input("Horas de uso por dia: "))
dias = int(input("Número de dias: "))
tarifa = float(input("Tarifa (R$/kWh): "))
orcamento = float(input("Orçamento (R$): "))

energia_kwh = potencia_w / 1000 * horas_dia * dias
custo = energia_kwh * tarifa

print(f"Consumo estimado: {energia_kwh:.2f} kWh")
print(f"Custo estimado: R$ {custo:.2f}")
if custo > orcamento:
    print("Custo acima do orçamento")
else:
    print("Custo dentro do orçamento")
````
