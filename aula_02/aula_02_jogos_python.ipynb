### Etapa 1 — Sortear o número (5 min)

O módulo `random` reúne funções de sorteio. Escrevemos `import random` uma vez e, depois, `random.randint(início, fim)` sorteia um inteiro entre os dois limites, **incluindo ambos**. Execute a célula algumas vezes e observe o número mudar.

````python
import random

numero = random.randint(0, 100)
print(numero)
````

### Etapa 2 — Um palpite e três respostas (7 min)

Um palpite pode estar **certo**, **abaixo** ou **acima** do número. `elif` significa “senão, se”: ele testa uma nova condição quando as anteriores foram falsas. O programa segue **somente o primeiro caminho verdadeiro**, e `else` cobre o que sobrou.

**Sua vez:** leia um palpite inteiro de 0 a 100 e mostre `Acertou!`, `Seu palpite ficou abaixo do número.` ou `Seu palpite ficou acima do número.`. Use o `numero` sorteado na etapa 1.

````python
# Escreva sua solução aqui.
````

<details><summary>Ver uma solução</summary>

```python
palpite = int(input("Seu palpite (0 a 100): "))
if palpite == numero:
    print("Acertou!")
elif palpite < numero:
    print("Seu palpite ficou abaixo do número.")
else:
    print("Seu palpite ficou acima do número.")
```
</details>

### Etapa 3 — Repetir com `for` e `range` (5 min)

`for` executa o bloco indentado uma vez para cada valor de uma sequência. `range(1, 11)` produz os inteiros de 1 a 10: **o início entra e o fim não entra**. A cada volta, a variável `tentativa` recebe o próximo valor.

Execute a célula abaixo. Quantas vezes a linha indentada é executada? E a última linha?

````python
for tentativa in range(1, 4):
    print(f"Tentativa {tentativa}")
print("Fim")
````

### Etapa 4 — Parar ao acertar com `break` (8 min)

`break` interrompe a repetição imediatamente, e o programa continua depois do bloco do `for`. Para saber, ao final, se o jogo terminou por acerto, guardamos um booleano: `acertou = False` antes do `for` e `acertou = True` no acerto. `not acertou` é verdadeiro quando `acertou` vale `False`.

<details><summary>Ver os passos para programar</summary>

1. Antes da repetição: sorteie o número e crie `acertou = False`.
2. Dentro de um `for` com dez voltas: leia o palpite e compare-o com o número.
3. No acerto: avise, marque `acertou = True` e use `break`.
4. Depois do `for`: se `not acertou`, mostre o número sorteado.
</details>

**Sua vez:** junte as etapas e escreva o jogo completo. Para testar, digite números inteiros de 0 a 100.

````python
# Escreva sua solução aqui.
````

<details><summary>Ver uma solução</summary>

```python
import random

numero = random.randint(0, 100)
acertou = False

for tentativa in range(1, 11):
    palpite = int(input(f"Tentativa {tentativa}/10 · seu palpite: "))
    if palpite == numero:
        print("Acertou!")
        acertou = True
        break
    elif palpite < numero:
        print("Seu palpite ficou abaixo do número.")
    else:
        print("Seu palpite ficou acima do número.")

if not acertou:
    print(f"Fim das tentativas. O número era {numero}.")
```
</details>

## 3. Par ou ímpar (15 min)

**Resto da divisão:** o operador `%` calcula o resto da divisão inteira. Um número é **par** quando `numero % 2 == 0` e **ímpar** quando `numero % 2 == 1`. Execute a célula e confira.

````python
print(7 % 2)
print(10 % 2)
print(10 % 2 == 0)
````

**Regras:**

- Você digita `par` ou `ímpar`, exatamente assim, em letras minúsculas.
- Depois, informa um inteiro de **0 a 10**. O computador sorteia outro número nesse intervalo.
- Se a paridade da **soma** for a que você escolheu, você vence; caso contrário, vence o computador.

**Exemplo de partida:**

```text
Par ou ímpar? par
Seu número (0 a 10): 3
Computador: 6 · Soma: 9 · Deu ímpar
O computador venceu.
```

**Pense antes de programar:** responda em português, sem escrever código.

1. O que o programa precisa calcular antes de decidir?
2. Como saber se um número é par sem consultar uma lista?
3. Precisamos testar cada combinação de números?

<details><summary>Conferir as respostas</summary>

1. A soma do seu número com o do computador.
2. Dividir por 2 e ver se sobra resto. Sem resto, é par; com resto 1, é ímpar.
3. Não. Basta comparar a paridade da soma com a escolha da pessoa.
</details>

<details><summary>Ver os passos para programar</summary>

1. Leia a escolha (texto) e o número (converta com `int()`).
2. Sorteie o número do computador e calcule a soma.
3. Use `%` para guardar `"par"` ou `"ímpar"` em uma variável `resultado`.
4. Compare `resultado` com a escolha usando `==`.
</details>

````python
# Escreva sua solução aqui.
````

```python
    resultado = "par"

    resultado = "ímpar"
```

## 4. Pedra, papel e tesoura (20 min)

**Regras:** uma rodada contra o computador. Você digita **1 = pedra**, **2 = papel** ou **3 = tesoura**; o computador sorteia um número de 1 a 3. Pedra quebra tesoura, papel embrulha pedra e tesoura corta papel. Escolhas iguais empatam.

| Você \ Computador | Pedra | Papel | Tesoura |
| --- | --- | --- | --- |
| Pedra | Empate | Derrota | **Vitória** |
| Papel | **Vitória** | Empate | Derrota |
| Tesoura | Derrota | **Vitória** | Empate |

**Combinando condições:** `and` junta duas comparações e só produz `True` quando **as duas** são verdadeiras. Por exemplo, `jogador == 1 and computador == 3` descreve pedra contra tesoura.

**Exemplo de partida:**

```text
1 = pedra · 2 = papel · 3 = tesoura
Sua escolha: 2
Você: 2 · Computador: 1
Papel embrulha pedra. Você venceu!
```

4. Precisamos descrever as derrotas uma a uma?

<details><summary>Conferir as respostas</summary>

1. Nove: três escolhas suas vezes três do computador.
2. As duas escolhas são iguais.
3. Pedra contra tesoura, papel contra pedra e tesoura contra papel.
4. Não. O que não é empate nem vitória só pode ser derrota.
</details>

<details><summary>Ver os passos para programar</summary>

1. Mostre as opções, leia a escolha e sorteie a do computador.
2. Mostre as duas escolhas.
3. Primeiro, trate o empate: as escolhas são iguais?
4. Escreva um `elif` com `and` para cada uma das três vitórias.
5. O que sobrar no `else` é derrota.
</details>

````python
# Escreva sua solução aqui.
````

<details><summary>Ver uma solução</summary>

```python

print(f"Você: {jogador} · Computador: {computador}")

    print("Empate!")
elif jogador == 1 and computador == 3:
    print("Pedra quebra tesoura. Você venceu!")
elif jogador == 2 and computador == 1:
    print("Papel embrulha pedra. Você venceu!")

    print("Tesoura corta papel. Você venceu!")

    print("O computador venceu.")
```

## 5. Aventura de decisões (15 min)

**Regras:** a estação espacial Órion perdeu energia. Você faz **duas escolhas** seguidas, digitando `1` ou `2`, e chega a um de **três finais**:

- Sala de comando → digitar o código: **missão cumprida**.
- Sala de comando → chamar a Terra: **resgate**.
- Laboratório → examinar a amostra: **descoberta**.
- Laboratório → ir à cápsula: **resgate**.

**Condições aninhadas:** a segunda pergunta depende da primeira resposta. Por isso, colocamos um `if` dentro do bloco de outro `if`, com mais um nível de indentação. Como só comparamos as respostas com os textos `"1"` e `"2"`, não precisamos convertê-las em número.

**Exemplo de partida:**

```text
Alerta! A estação Órion perdeu energia.
1 = sala de comando, 2 = laboratório: 2
1 = examinar a amostra, 2 = ir à cápsula: 1
A amostra brilha e gera energia. Descoberta!
```

**Pense antes de programar:** responda em português, sem escrever código.

1. Quantas perguntas a pessoa responde em cada partida?
2. A segunda pergunta é sempre a mesma?
3. Quantos caminhos existem? E quantos finais?

<details><summary>Conferir as respostas</summary>

1. Duas.
2. Não. Ela depende da resposta à primeira.
3. Quatro caminhos e três finais: dois caminhos levam ao resgate.
</details>

<details><summary>Ver os passos para programar</summary>

1. Mostre a situação e leia a primeira escolha.
2. No bloco de `if primeira == "1":`, faça a pergunta da sala de comando e decida o final.
3. No bloco de `else`, faça a pergunta do laboratório e decida o final.
</details>

````python
# Escreva sua solução aqui.
````

<details><summary>Ver uma solução</summary>

```python

primeira = input("1 = sala de comando, 2 = laboratório: ")

if primeira == "1":
    segunda = input("1 = digitar o código, 2 = chamar a Terra: ")
    if segunda == "1":
        print("Os motores voltam a funcionar. Missão cumprida!")
    else:
        print("Uma nave vem buscar você. Resgate!")
else:
    segunda = input("1 = examinar a amostra, 2 = ir à cápsula: ")
    if segunda == "1":
        print("A amostra brilha e gera energia. Descoberta!")
    else:
        print("A cápsula leva você de volta à Terra. Resgate!")
```
</details>

## 6. Quiz de multiplicação (20 min)

**Regras:** o programa faz **cinco perguntas**. Em cada uma, sorteia dois fatores de **1 a 10** e pede o produto. Se a resposta estiver certa, avisa e soma um ponto; se estiver errada, mostra a resposta correta. No fim, apresenta a pontuação.

**Contador:** uma variável que começa em zero, **antes** da repetição, e aumenta a cada acerto com `pontos = pontos + 1`.

**Exemplo de partida (trecho):**

```text
Pergunta 1: 7 x 8 = 56
Certo!
Pergunta 2: 6 x 9 = 56
Errado. A resposta é 54.
...
Pontuação final: 4 de 5
```

**Pense antes de programar:** responda em português, sem escrever código.

1. O que acontece em cada uma das cinco perguntas?
2. O que acontece uma vez só, antes e depois das perguntas?
3. Como o programa lembra quantas respostas estavam certas?

<details><summary>Conferir as respostas</summary>

1. Sortear dois fatores, perguntar o produto, conferir e avisar se está certo.
2. Antes, os pontos começam em zero. Depois, mostramos a pontuação final.
3. Com um contador que aumenta um ponto a cada acerto.
</details>

<details><summary>Ver os passos para programar</summary>

1. Crie `pontos = 0` antes do `for`.
2. Repita cinco vezes: sorteie os dois fatores e leia a resposta como inteiro.
3. Compare a resposta com o produto e atualize `pontos` quando acertar.
4. Depois do `for`, mostre a pontuação final.
</details>

````python
# Escreva sua solução aqui.
````

<details><summary>Ver uma solução</summary>

```python
import random

pontos = 0
for pergunta in range(1, 6):
    fator1 = random.randint(1, 10)
    fator2 = random.randint(1, 10)
    resposta = int(input(f"Pergunta {pergunta}: {fator1} x {fator2} = "))
    if resposta == fator1 * fator2:
        print("Certo!")
        pontos = pontos + 1
    else:
        print(f"Errado. A resposta é {fator1 * fator2}.")

print(f"Pontuação final: {pontos} de 5")
```
</details>
